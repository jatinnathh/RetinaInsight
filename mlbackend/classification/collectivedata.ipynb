```python
import pandas as pd 
from pathlib import Path
import shutil
from tqdm import tqdm 
```

```python
BASE_PATH=Path(r'../data/classification')

APTOS_IMAGES=BASE_PATH/'aptos2019/train_images'
APTOS_LABELS=BASE_PATH/'aptos2019/train.csv'
IDRID_IMAGES = BASE_PATH / "archive (6)/imagenes/imagenes"
IDRID_LABELS = BASE_PATH / "archive (6)/idrid_labels.csv"
FINAL_DATA = Path("final_data")
print("APTOS Images:", APTOS_IMAGES)
print("APTOS Labels:", APTOS_LABELS)

print("IDRiD Images:", IDRID_IMAGES)
print("IDRiD Labels:", IDRID_LABELS)

print("Final Data:", FINAL_DATA)
```

```python
print("APTOS images exist ", APTOS_IMAGES.exists())
print("APTOS labels exist ", APTOS_LABELS.exists())
print("IDRID images exist ", IDRID_IMAGES.exists())
print("IDRID labels exist ", IDRID_LABELS.exists())
```

```python
label_map = {
    0: "no_dr",
    1: "mild",
    2: "moderate",
    3: "severe",
    4: "proliferative_dr"
}
```

```python
FINAL_DATA.mkdir(exist_ok=True)

for folder_name in label_map.values():
    (FINAL_DATA/folder_name).mkdir(exist_ok=True)
print("created subflder successfully ")
```

```python
aptos_df=pd.read_csv(APTOS_LABELS)
print(aptos_df.head())
print("\nColumns:")
print(aptos_df.columns)

print("\n class distrivurion ")

print(aptos_df['diagnosis'].value_counts().sort_index())
```

```python
copied=0
missing=0

for _ , row in tqdm(aptos_df.iterrows(),total=len(aptos_df)):
    imageId=str(row['id_code'])
    label=int(row['diagnosis'])

    className=label_map[label]

    sourceImg=APTOS_IMAGES/f"{imageId}.png"

    destination=FINAL_DATA/className/f"aptos_{imageId}.png"

    if sourceImg.exists():
        shutil.copy2(sourceImg,destination)
        copied+=1
    else:
        missing+=1

print("APTOS iamge copying finished")
print(" total copies values ",copied)
print("total misssingiamges ",missing)
```

```python
idrid_df=pd.read_csv(IDRID_LABELS)

# print(idrid_df.head())
idrid_df=idrid_df[['id_code','diagnosis','Risk of macular edema ']]
print("\nColumns:")
print(idrid_df.columns)

print("\nClass distribution:")
print(idrid_df["diagnosis"].value_counts().sort_index())
```

```python

copies,missing=0,0

for _ , row in tqdm(idrid_df.iterrows(),total=len(idrid_df)):
    label=row[1]
    imageId=row[0]

    className=label_map[label]

    sourceImg=IDRID_IMAGES/f"{imageId}.jpg"
    destination=FINAL_DATA/className/f"idrid_{imageId}.jpg"
    if sourceImg.exists():
        shutil.copy2(sourceImg,destination)
        copies+=1
    else:
        missing+=1

print("copieed ",copies)
print("missing ",missing)
    
```
